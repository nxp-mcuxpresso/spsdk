# MCXN9xx Code Read Protection (CRP)

This notebook shows how to read the Code Read Protection (CRP) level of an MCXN9xx device (e.g. `mcxn946`/`mcxn947`) using the SPSDK *nxpdebugmbox* application.

Code Read Protection restricts read/write access to the on-chip flash and debug resources. The current CRP level is exposed through the debug mailbox and can be queried with the `nxpdebugmbox cmd get-crp` command. This is useful to verify the security state of a device during development and provisioning. The same *nxpdebugmbox* tool is used for [debug authentication](./mcxn9xx_debug_auth.ipynb) on this device.

<div class="alert alert-block alert-warning">
<b>WARNING:</b> This example only reads the CRP level; it does not change any device configuration. Raising the CRP level on a real device (via the CMPA) can permanently restrict debug access.
</div>

## 1. Prerequisites

- SPSDK is needed with examples extension. `pip install spsdk[examples]` (Please refer to the [installation](../../_knowledge_base/installation_guide.rst) documentation.)
- This example uses an MCXN9xx board (e.g. FRDM-MCXN947) connected via a debug probe (pyOCD / MCU-Link / J-Link).

```python
# This env variable sets colored logger output to STDOUT
%env JUPYTER_SPSDK=1
# Set a magic for command execution and echo
%alias execute echo %l && %l
%alias_magic ! execute
```

```python
FAMILY = "mcxn946"  # change this to your MCXN9xx family (e.g. mcxn947)
INTERFACE = "jlink"  # debug probe interface: pyocd, mcu-link or jlink
VERBOSITY = (
    ""  # verbosity of commands, might be -v or -vv for debug or blank for no additional info
)
```

## 2. Read the CRP level

The CRP level is read through the debug mailbox in three steps:

1. `start` opens a debug mailbox session,
2. `get-crp` reads the current CRP level (it must be called after `start` and with the `-n`/`--no-reset` option so the session is preserved),
3. `exit` closes the debug mailbox session.

```python
# Open the debug mailbox session
%! nxpdebugmbox $VERBOSITY -i $INTERFACE cmd -f $FAMILY start
```

```python
# Read the current CRP level (no-reset keeps the mailbox session open)
%! nxpdebugmbox $VERBOSITY -n -i $INTERFACE cmd -f $FAMILY get-crp
```

```python
# Close the debug mailbox session
%! nxpdebugmbox $VERBOSITY -i $INTERFACE cmd -f $FAMILY exit
```
